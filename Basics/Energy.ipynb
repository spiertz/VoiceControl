# Energy

The energy of a time continuous signal $x(t)$ is evaluated by:

$E=\int_0^T x^2(t) dt$

In computer based signal processing, the signals are typically time discrete. The energy of a digital signal $x(n)$ of length $N$ is defined by:

$E=\sum_{n=0}^{N-1} x^2(n)$

```python
import numpy as np

gain = np.random.rand(1) * 10 
x = np.random.randn(10000) * gain

def EvalEnergy(x):
    return np.sum(x**2)

E = EvalEnergy(x)
assert np.sqrt(np.abs(E / x.shape[0] - gain**2)) < 0.5, f"Expected energy {x.shape[0]* (gain**2)}, but got {E}"
```
